Read in the citations and patents data and check that the data makes sense. Note that unlike in the RDD solution, the data is automatically inferred to be Integer() types.

```python
citations = spark.read.load('cite75_99.txt.gz',
            format="csv", sep=",", header=True,
            compression="gzip",
            inferSchema="true")
```

```python
citations.show(5)
```

```python
patents = spark.read.load('apat63_99.txt.gz',
            format="csv", sep=",", header=True,
            compression="gzip",
            inferSchema="true")
```

```python
patents.show(5)
```

```python
#aliases for patent data joins
citing_patent = patents.alias("citing_patent")
cited_patent = patents.alias("cited_patent")

#join patent data for citing and cited
step = citations.join(citing_patent, citations.CITING==citing_patent.PATENT, how="inner").join(cited_patent, citations.CITED==cited_patent.PATENT, how="inner")

#drop unnecessary columns and filter for same state citations
step = step.select(
    "CITING",
    "CITED",
    col("citing_patent.POSTATE").alias("CITING_STATE"),
    col("cited_patent.POSTATE").alias("CITED_STATE")
).filter(
    col("CITING_STATE") == col("CITED_STATE")
)

#group on CITING and get counts
step = step.groupby("CITING").count()

#rejoin intermediate counts table to patents, sort in descending order, limit to 10
final = patents.join(step,patents.PATENT==step.CITING,how="inner").sort("count",ascending=False).limit(10)

#display results
final.show()
```
